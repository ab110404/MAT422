3.3.1. Necessary and sufficient conditions of local minimizers
- In this example, I will analyze f(x) = x^4 - 4x^2 + 3 using derivatives and Python.

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
  return x**4 - 4*x**2 + 3
def f_prime(x):
  return 4*x**3 - 8*x
def f_second(x):
  return 12*x**2 - 8
print("Function: f(x) = x^4 - 4x^2 + 3")
print("First derivative: f'(x) = 4x^3 - 8x")
print("Second derivative: f''(x) = 12x^2 - 8")
```

- For this function: f'(x) = 4x^3 - 8x. Setting the derivative equal to zero gives: 4x(x^2 - 2) = 0. Therefore, the critical points are: x=0, x=-sqrt(2), and x=sqrt(2).
- I will now evaluate the second derivative at each critical point to determine whether it is a minimum or maximum.

```python
critical_points = np.array([-np.sqrt(2), 0, np.sqrt(2)])
print("Critical points:")
for  x in critical_points:
  print(f"x = {x:.4f}")
```

- The second derivative helps determine whether a critical point is a local minimum or maximum.
- So I will now apply the second derivative test to each critical point.

```python
for x in critical_points:
  second = f_second(x)

  if second > 0:
    classification = "Local minimum"
  elif second < 0:
    classification = "Local maximum"
  else:
    classification = "Inconclusive"
  print(f"x = {x:.4f}, f''(x) = {second:.4f}: {classification}")
```

- I will evaluate the function at the critical points and compare the results.
- Since f(x) = (x^2 - 2)^2 - 1, the function is always greater than or equal to -1.
- This allows us to identify the global minimum exactly.

```python
for x in critical_points:
  print(f"x = {x:.4f}, f(x) = {f(x):.4f}")
minimum_value = np.min(f(critical_points))
print("\nGlobal minimum value:", minimum_value)
print("Global minimizers: x = +-sqrt(2)")
```

- I will plot the function and mark the critical points. The graph will help visualize the local minima, the local maximum, and the global minimum values.

```python
x = np.linspace(-2.5, 2.5, 400)
y = f(x)
plt.figure(figsize=(8, 5))
plt.plot(x, y, label="f(x) = x^4 - 4x^2 + 3")
plt.scatter(
    [-np.sqrt(2), np.sqrt(2)],
    [-1, -1],
    color="red",
    label="Global minima",
    zorder=3
)
plt.scatter(
    [0], [3],
    color="green",
    label="Local maximum",
    zorder=3
)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Local and Global Minimizers")
plt.legend()
plt.grid(True)
plt.show()
```

- In this section, I used Python to identify and classify the critical points of a function. The first derivative was used to find the critical points, and the second derivative was used to determine whether each point was a local minimum or maximum. These results demonstrate how necessary and sufficient conditions can be used to solve optimixation problems.

3.3.2. Convexity and global minimizers
- In this example, I will analyze the function f(x) = x^2 - 4x + 5 using Python

```python
def f(x):
  return x**2 - 4*x + 5
def f_prime(x):
  return 2*x - 4
def f_second(x):
  return 2
print("Function: f(x) = x^2 - 4x + 5")
print("First derivative: f'(x) = 2x -4")
print("Second derivative: f''(x) = 2")
```

- For the function f(x) = x^2 - 4x +5: f''(x) = 2. Since the second derivative is positive everywhere, the function is strictly convex. This means that any local minimizer is also a global minimizer.

```python
x_values = np.linspace(-2, 6, 100)
second_derivatives = np.array(
    [f_second(x) for x in x_values]
)
if np.all(second_derivatives > 0):
  print("The function is strictly convex.")
else:
  print("Strict convexity was not verified.")
```

- To find the global minimizer, I will set the first derivative equal to zero.
f'(x) = 2x - 4 = 0. Which gives x=2. Since the function is strictly convex, this critical point is the unique global minimizer.

```python
x_min = 4 / 2
y_min = f(x_min)
print("Global minimizer:", x_min)
print("Minimum function value:", y_min)
print("Second derivative:", f_second(x_min))
```

- I will graph the function to visualize its convex shape and identify the global minimum.

```python
x = np.linspace(-2, 6, 400)
y = f(x)
plt.figure(figsize=(8, 5))
plt.plot(x, y, label="f(x) = x^2 - 4x + 5")
plt.scatter(
    x_min, y_min,
    color="red",
    label="Global minimum",
    zorder=3
)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Convex Function and Global Minimizer")
plt.legend()
plt.grid(True)
plt.show()
```

- I will choose two points and use alpha = 0.5 to test the convexity inequality.

```python
x1 = 0
x2 = 4
alpha = 0.5
left_side = f((1-alpha)*x1 + alpha*x2)
right_side = (1-alpha)*f(x1) + alpha*f(x2)
print("Left side:", left_side)
print("Right side:", right_side)
print("Convexity inequality satisfied:", left_side <= right_side)
```

Conclusion:
- In this section, I explored convexity and global minimizers using python. This example demonstrates why convexity is useful in optimization: a local minimizer of a convex function is guaranteed to be a global minimizer.

3.3.3. Gradient descent
- In this example, I will use gradient descent to minimize f(x) = x^2 - 4x + 5.

```python
def f(x):
  return x**2 - 4*x + 5
def gradient(x):
  return 2*x - 4
print("Function: f(x) = x^2 - 4x + 5")
print("Gradient: f'(x) = 2x -4")
```

- I will start with an initial guess of x = 0. The learning rate will be 0.1, and I will perform 20 iterations. The goal is to move closer to the minimum at x=2.

```python
x = 0.0
alpha = 0.1
iterations = 20
history = [x]
for i in range(iterations):
  x = x - alpha * gradient(x)
  history.append(x)
  print(f"Iteration {i+1}: x = {x:.4f}, f(x) = {f(x):.4f}")
```

- The exact minimum can be found by setting the first derivative equal to zero, f'(x) = 2x - 4 = 0, which gives x=2. Since the function is strictly convex, this is the unique global minimizer. I will now compare the gradient descent approximation with the exact solution.

```python
exact_minimum = 2.0
print("Gradient descent result:", round(x, 4))
print("Exact minimizer:", exact_minimum)
print("Approximation error:", round(abs(x - exact_minimum), 4))
print("Minimum function value:", round(f(x), 4))
```

- I will graph the function and the points visited during gradient descent. The graph will show how the algorithm starts at x=0 and gradually moves toward the global minimum at x=2.

```python
x_values = np.linspace(-1, 5, 400)
y_values = f(x_values)
plt.figure(figsize=(8, 5))
plt.plot(x_values, y_values, label="f(x)")
plt.scatter(
    history,
    [f(value) for value in history],
    color="red",
    label="Gradient descent steps"
)
plt.scatter(
    2, 1,
    color="green",
    s=100,
    label="Exact minimum",
    zorder=5
)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Gradient Descent Optimization")
plt.legend()
plt.grid(True)
plt.show()
```

- I will compare different learning rates to see how they affect convergence.

```python
learning_rates = [0.05, 0.1, 0.3, 1.1]
for alpha in learning_rates:
  x_current = 0.0
  for i in range(20):
    x_current = x_current - alpha * gradient(x_current)
  print(
      f"Learning rate = {alpha}: "
      f"x = {x_current:.4f}, "
      f"f(x) = {f(x_current):.4f}"
  )
```

Conclusion:
- In this section, I explored gradient descent as a numerical optimization method. This example demonstrates how gradient descent can approximate the minimum of a function without directly solving the derivative equation.
