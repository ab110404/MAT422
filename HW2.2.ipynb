2.2.1 Probability Axioms
- Probability is used to describe experiments whose outcomes are uncertain.
- In this example, I will use the experiment of rolling a fair six-sided die to demonstrate sample spaces, events, and the basic probability axioms.
- Suppose the experiment is rolling one fair six-sided die. The sample space is : S = {1, 2, 3, 4, 5, 6}. A simple event contains exactly one outcome. For example: A = {4}. Then a compound event contains more than one outcome. The event of rolling an even number is B = {2, 4, 6}.

```python
import numpy as np
```

```python
S = {1, 2, 3, 4, 5, 6}
A = {4}
B = {2, 4, 6}
print("Sample space S =", S)
print("Simple event A =", A)
print("Compound event B =", B)
```

- For a fair six-sided die, each outcome has probability 1/6. For an event E, its probability can be calculated as P(E) = number of outcomes in E / number of outcomes in S. Now I will create a function to calculate the probability of an event.

```python
def probability(event, sample_space):
  return len(event) / len(sample_space)

P_A = probability(A, S)
P_B = probability(B, S)
print("P(A) =", P_A)
print("P(B) =", P_B)
```

- The first probability axiom states that the probability of any event A must satisfy 0 <= P(A) <= 1. I will verify this using the compound event B = {2, 4, 6}.

```python
P_B = probability(B, S)
print("P(B) =", P_B)
print("Is 0 <= P(B) <= 1?", 0 <= P_B <= 1)
```

- The result is true, so the first probability axiom is satisfied.
- The second probability axiom states that P(S) = 1. Since the sample space contains every possible outcome, one of these outcomes must occur when the experiment is performed.

```python
P_S = probability(S, S)
print("P(S) =", P_S)
print("Does P(S) = 1?", P_S == 1)
```

- The result is true, which verifies the second probability axiom.
- For mutually exclusive events A and B, P(A union B) = P(A) + P(B). Now, we will consider the events C = {1, 3, 5}, representing an old roll, and D = {2, 4, 6}, representing an even roll.

```python
C = {1, 3, 5}
D = {2, 4, 6}
print("C =", C)
print("D =", D)
print("\nIntersection of C and D:", C.intersection(D))
```

```python
P_C = probability(C, S)
P_D = probability(D, S)
union = C.union(D)
P_union = probability(union, S)
print("P(C) =", P_C)
print("P(D) =", P_D)
print("\nP(C union D) =", P_union)
print("P(C) + P(D) =", P_C + P_D)
print("\nAre they equal?", P_union == P_C + P_D)
```

Conclusion:
- The calculations verified the probability of an event is between 0 and 1. The probability of the entire sample space is 1. For mutually exclusive events, the probability of their union equals the sum of their individual probabilities. These results demonstrate the basic probability axioms.

2.2.2 Conditional Probability

- In this example, I will use a fair six-sided die to demonstrate conditional probability, the multiplication rule, and independence.
- The sample space is S = {1, 2, 3, 4, 5, 6}. Let A = {2, 4, 6}, the event that an even number is rolled and B = {4, 5, 6}, the event that a number greater than 3 is rolled. The intersection A ∩ B contains the outcomes that belong to both events.

```python
S = {1, 2, 3, 4, 5, 6}
A = {2, 4, 6}
B = {4, 5, 6}
intersection = A.intersection(B)
print("Sample space S =", S)
print("Event A =", A)
print("Event B =", B)
print("A intersection B =", intersection)
```

- The conditional probability formula is P(A|B) = P(A ∩ B) / P(B). The die is fair, probabilities can be calculated by dividing the number of outcomes in an event by the total number of outcomes.

```python
def probability(event, sample_space):
  return len(event) / len(sample_space)

P_A = probability(A, S)
P_B = probability(B, S)
P_intersection = probability(intersection, S)
P_A_given_B = P_intersection / P_B
print("P(A) =", P_A)
print("P(B) =", P_B)
print("P(A intersection B) =", P_intersection)
print("P(A|B) =", P_A_given_B)
```

- Conditional probability gives the multiplication rule P(A ∩ B) = P(A|B)P(B). I will verify that both sides of this equation produce the same result.

```python
left_side = P_intersection
right_side = P_A_given_B * P_B
print("P(A intersection B) =", left_side)
print("P(A|B) * P(B) =", right_side)
print("\nAre they equal?", np.isclose(left_side, right_side))
```

- Two events A and B are independent if P(A|B) = P(A), or equivalently, P(A ∩ B) = P(A)P(B).
- C = {2, 4, 6}, the event of rolling an even number.
- D = {1, 2, 3}, the event of rolling a number less than or equal to 3.

```python
C = {2, 4, 6}
D = {1, 2, 3}

C_intersection_D = C.intersection(D)

P_C = probability(C, S)
P_D = probability(D, S)
P_C_intersection_D = probability(C_intersection_D, S)

print("C =", C)
print("D =", D)
print("C intersection D =", C_intersection_D)
print("\nP(C) =", P_C)
print("P(D) =", P_D)
print("P(C intersection D) =", P_C_intersection_D)
```

```python
C = {2, 4, 6}
D = {1, 2, 4, 5}
C_intersection_D = C.intersection(D)

P_C = probability(C, S)
P_D = probability(D, S)
P_C_intersection_D = probability(C_intersection_D, S)

print("C =", C)
print("D =", D)
print("C intersection D =", C_intersection_D)
print("\nP(C) =", P_C)
print("P(D) =", P_D)
print("P(C intersection D) =", P_C_intersection_D)
```

- To determine whether C and D are independent, I will check whether P(C ∩ D) = P(C)P(D).

```python
product = P_C * P_D
print("P(C intersection D) =", P_C_intersection_D)
print("P(C) * P(D) =", product)
print("\nAre C and D independent?", np.isclose(P_C_intersection_D, product))
```

Conclusion:
- This example demonstrated conditional probability using outcomes from a fair six-sided die.

2.2.3 Discrete Random Variables

- In this example, I will toss two coins and define the random variable X as the number of heads obtained.
- The sample space is S = {HH, HT, TH, TT}. Define the random variable X as the number of heads obtained.
- Therefore, X(HH)=2, X(HT)=1, X(TH)=1, X(TT)=0.

```python
S = ["HH", "HT", "TH", "TT"]

def X(outcome):
  return outcome.count("H")
for outcome in S:
  print(outcome, "->", X(outcome))
```

- The function X maps every outcome in the sample space to a number. The possible values of X are {0, 1, 2}. Since the possible values form a finite set, X is a discrete random variable.

```python
possible_X = sorted(set(X(outcome) for outcome in S))
print("Possible values of X:", possible_X)
```

- Since the two coins are fair, the four outcomes HH, HT, TH, and TT are equally likely. For this experiment: P(X = 0) = 1/4, P(X = 1) = 2/4 = 1/2, P(X = 2) = 1/4

```python
for x in possible_X:
  count = sum(X(outcome) == x for outcome in S)
  probability = count / len(S)
  print("P(X =", x, ") =", probability)
```

- For a valid discrete probability distribution, every probability must be between 0 and 1, and the probabilities of all possible values must add to 1.

```python
probabilities = []
for x in possible_X:
  count = sum(X(outcome) == x for outcome in S)
  probabilities.append(count / len(S))
print("Probabilities:", probabilities)
print("Sum of probabilities =", sum(probabilities))
```

- The expected value of a discrete random variable represents its long-run average value. It can be calculated using this formula, E[X] = ∑ xP(X = x).

```python
expected_value = sum(
    x * probabilities[i]
    for i, x in enumerate(possible_X)
)
print("E[X] =", expected_value)
```

Conclusion:
- In this example, two fair coins were tossed and the random variable X was defined as the number of heads obtained. The probabilities sum to 1, and the expected value of X is 1. This example demonstrates how a discrete random variable maps outcomes of a random experiment to numerical values.

2.2.4 Continuous Random Variables
- In this example, I will use a uniform continuous random variable on the interval [0, 10].
- Suppose X can take any value between 0 and 10 with equal likelihood. The probability density function is f(x) = 1/10 for 0 <= x <= 10, and f(x) = 0 otherwise.

```python
import matplotlib.pyplot as plt
def f(x):
  if 0 <= x <= 10:
    return 1/10
  else:
    return 0
print("f(2) =", f(2))
print("f(5) =", f(5))
print("f(12) =", f(12))
```

- A valid probability density function must have a total area of 1. Therefore, I will calculate ∫ f(x) dx over the interval from 0 to 10.

```python
from scipy.integrate import quad
total_area, error = quad(f, 0, 10)
print("Total area under f(x) =", total_area)
```

- For a continuous random variable, probability is calculated as the area under the probability density function over an interval. Now I will calculate, P(2 <= X <= 6). Using the probability density function, P(2 <= X <= 6) = ∫ f(x) dx from 2 to 6.

```python
probability, error = quad(f, 2, 6)
print("P(2 <= X <= 6) =", probability)
```

- The probability can also be represented graphically. The shaded region between x = 2 and x = 6 represents the probability of 0.4.

```python
x = np.linspace(0, 10, 500)
y = np.ones_like(x) * 0.1

plt.plot(x, y, label="f(x)")
x_fill = np.linspace(2, 6, 200)
y_fill = np.ones_like(x_fill) * 0.1
plt.fill_between(x_fill, y_fill, alpha=0.3)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Uniform Probability Density Function")
plt.ylim(0, 0.15)
plt.legend()
plt.show()
```

- An important property of a continuous random variable is that X being exactly one specific value is 0. For example, P(X = 5) = 0.

```python
single_point_probability, error = quad(f, 5, 5)
print("P(X = 5) =", single_point_probability)
```
