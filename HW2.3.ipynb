2.3.1 Joint probability distributions

- A joint probability distribution describes the probabilities associated with two or more random variables.
- The example I will be using, suppose X can take the values 0 or 1 and Y can take the values 0 or 1. The following joint probability distribution will be used:
x: 0,0,1,1
y: 0,1,0,1
p(x,y): 0.10,0.20,0.30,0.40

```python
import numpy as np
```

```python
joint_pmf = {
    (0, 0): 0.10,
    (0, 1): 0.20,
    (1, 0): 0.30,
    (1, 1): 0.40
}

for (x, y), p in joint_pmf.items():
  print("X =", x, "Y =", y, "p(x,y) =", p)
```

- The joint PMF assigns a probability to each possible pair of values (X, Y).
- For example, P(X = 1 and Y = 0) = 0.30.
- A joint probability mass function must satisfy p(x,y) >= 0 for every possible pair (x,y). I will verify that none of the probabilities in the joint distribution are negative.

```python
nonnegative = all(p >= 0 for p in joint_pmf.values())
print("Are all probabilities nonnegative?", nonnegative)
```

- For a valid joint probability mass function, the sum of all possible joint probabilities must equal 1. Now I will add values of p(x,y) to verify this property.

```python
total_probability = sum(joint_pmf.values())
print("Sum of all joint probabilities =", total_probability)
```

The marginal probability distribution of X is found by summing the joint probabilities over all possible values of Y.

```python
x_values = [0, 1]
y_values = [0, 1]
marginal_X = {}

for x in x_values:
  marginal_X[x] = sum(joint_pmf[(x, y)] for y in y_values)
print("Marginal distribution of X:")
for x, p in marginal_X.items():
  print("P(X =", x, ") =", p)
```

- The marginal probability distribution of Y is found by summing the joint probabilities over all possible values of X.

```python
marginal_Y = {}

for y in y_values:
  marginal_Y[y] = sum(joint_pmf[(x, y)] for x in x_values)
print("Marginal distribution of Y:")

for y, p in marginal_Y.items():
  print("P(Y =", y, ") =", p)
```

Conclusion:
- This example demonstrated a joint probability distribution for two discrete random variables X and Y.
- The joint PMF was verified to contain only nonnegative probabilities.
- The marginal distributions were then calculated by summing the joint probabilities over the values of the other variable.

2.3.2. Correlation and dependence

- Covariance measures how two random variables vary together.
- I will first create two variables that increase together. Based on their behavior, I expect their covariance and correlation to be positive.

```python
X = np.array([1, 2, 3, 4, 5])
Y = np.array([2, 4, 5, 7, 9])

covariance = np.cov(X, Y, ddof=0)[0, 1]
correlation = np.corrcoef(X, Y)[0, 1]
print("Covariance =", covariance)
print("Correlation =", correlation)
```

- Next, I will use variables where X increases while Y decreases. I expect the covariance and correlation to be negative.

```python
X = np.array([1, 2, 3, 4, 5])
Y = np.array([10, 8, 7, 5, 2])

covariance = np.cov(X, Y, ddof=0)[0, 1]
correlation = np.corrcoef(X, Y)[0, 1]
print("Covariance =", covariance)
print("Correlation =", correlation)
```

- A covariance near zero indicates that there is little linear relationship between the two variables. I will now create an example where Y does not consistently increase or decrease as X increases.

```python
X = np.array([1, 2, 3, 4, 5])
Y = np.array([3, 1, 4, 1, 3])

covariance = np.cov(X, Y, ddof=0)[0, 1]
correlation = np.corrcoef(X, Y)[0, 1]
print("Covariance =", covariance)
print("Correlation =", correlation)
```

- In the textbook, it states that the covariance of a random variable with itself is equal to its variance: Cov(X,X) = Var(X). I will now verify this.

```python
X = np.array([1, 2, 3, 4, 5])
cov_xx = np.cov(X, X, ddof=0)[0, 1]
var_x = np.var(X)
print("Cov(X,X) =", cov_xx)
print("Var(X) =", var_x)
print("Are they equal?", np.isclose(cov_xx, var_x))
```

Conclusion:
- These examples demonstrate how covariance and correlation describe the relationship between two random variables.
- A positive covariance occurs when the variables tend to increase together, while a negative covariance occurs when one tends to decrease as the other increases.
- Then the final example verified that the covariance of a random variable with itself is equal to its variance.

2.3.3. Random samples
- I will use a normal population with mean 𝜇 = 50 and standard deviation 𝜎 = 10. But first, I will generate one random sample of size n = 25 and calculate its sample mean.

```python
np.random.seed(42)
mu = 50
sigma = 10
n = 25
sample = np.random.normal(mu, sigma, n)
print("Random sample:")
print(sample)
print("\nSample mean =", np.mean(sample))
```

- The textbook sattes that E(X) = 𝜇. This means that if many random samples are taken, the average of their sample means should approach the population mean. I will generate 10,000 random samples and calculate the mean of each sample.

```python
num_samples = 10000
sample_means = []
for i in range(num_samples):
  sample = np.random.normal(mu, sigma, n)
  sample_means.append(np.mean(sample))
mean_of_sample_means = np.mean(sample_means)
print("Population mean =", mu)
print("Mean of sample means =", mean_of_sample_means)
```

For this example, 𝜎 = 10 and n = 25. Therefore, the theoretical standard deviation of the sample mean is 10/ square root 25 = 2. I will compare this theoretical value with the standard deviation of the simulated sample means.

```python
theoretical_std = sigma / np.sqrt(n)
simulated_std = np.std(sample_means)
print("Theoretical standard deviation =", theoretical_std)
print("Simulated standard deviation =", simulated_std)
```

- The formula 𝜎_X = 𝜎/ square root n shows that increasing the sample size decreases the standard deviation of the sample mean. So, now I will compare the theoretical standard deviation for several different sample sizes.

```python
sample_sizes = [4, 25, 100]
for n_value in sample_sizes:
  std_mean = sigma / np.sqrt(n_value)
  print("n =", n_value, "Standard deviation =", std_mean)
```

Conclusion:
- This example demonstrated the properties of a simple random sample and the sampling distribution of the sample mean. By increasing the sample size, it decreased the variability of the sample mean. Overall, every example was shown for HW 2.3
