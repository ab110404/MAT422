1.2.1 Linear Spaces

```python
###A linear combination is a new vector formed by multiplying vectors by scalars and then adding the results. If w1 and w2 are vectors, then a linear combination has the form a1w1 + a2w2, where a1 and a2 are real numbers.In this example, I will use python to form a linear combination of two vectors and verify that the resulting vector belongs to their span.

import numpy as np
```

```python
###Let w1 = (1, 2) and w2 = (3, 1).Choose the scalars a1 = 2 and a2 = -1. The linear combination is u = a1w1 + a2w2

w1 = np.array([1, 2])
w2 = np.array([3, 1])

alpha1 = 2
alpha2 = -1

print("w1 =", w1)
print("w2 =", w2)
print("alpha1 =", alpha1)
print("alpha2 =", alpha2)
```

```python
u = alpha1 * w1 + alpha2 * w2

print("Linear combination u =", u)
```

```python
###The result of the linear combination is u = (-1, 3). Since u was created using scalar multiples of w1 and w2 and then adding them together, u belongs to span(w1, w2).

A = np.column_stack((w1, w2))

coefficients = np.linalg.solve(A, u)

print("Coefficients =", coefficients)

###Python found the coefficients c1 = 2 and c2 = -1. Therefore, u = 2w1 - w2. This shows that u is a linear combination of w1 and w2, so u belongs to span(w1, w2).
```

1.2.2 Orthogonality

```python
### Two vectors are orthogonal if their inner product, or dot product, is equal to zero. For vectors u and v, <u,v> = u * v. The norm of a vector measures its length and is defined by ||u|| = sqrt(u1^2 + u2^2 + ...+ un^2).
u = np.array([3, 4], dtype=float)
v = np.array([-4, 3], dtype=float)

print("u =", u)
print("v =", v)
```

```python
### To determine whether u and v are orthogonal, I will calculate their dot product

dot_product = np.dot(u, v)
print("Dot product =", dot_product)
```

```python
### The norm of a vector represents its length.

norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)
print("Norm of u =", norm_u)
print("Norm of v =", norm_v)
```

```python
### The norms of both vectors are 5. The vectors are orthogonal, but they are not yet orthonormal because an orthogonal vector must have a norm of 1.

u_hat = u / norm_u
v_hat = v / norm_v
print("Normalized u =", u_hat)
print("Normalized v =", v_hat)
```

```python
### For the vectors to be orthonormal: their dot product must equal 0 and each vector must have a norm of 1

print("Dot product =", np.dot(u_hat, v_hat))
print("Norm of normalized u =", np.linalg.norm(u_hat))
print("Norm of normalized v =", np.linalg.norm(v_hat))
```

```python
### For orthogonal vectors u and v, the Pythagorean theorem states that ||u + v||^2 = ||u||^2 + ||v||^2

left_side = np.linalg.norm(u + v)**2
right_side = np.linalg.norm(u)**2 + np.linalg.norm(v)**2
print("||u + v||^2 =", left_side)
print("||u||^2 + ||v||^2 =", right_side)

### The vectors are orthogonal
```

1.2.3 Gram_Schmidt process

```python
### In this example, I will apply the Gram_Schmidt process to a1 = (1, 1, 0) and a2 = (1, 0, 1)

a1 = np.array([1, 1, 0], dtype=float)
a2 = np.array([1, 0, 1], dtype=float)
print("a1 =", a1)
print("a2 =", a2)
```

```python
### The first orthonormal vector is obtained by normalizing a1, q1 = a1 / ||a1||

q1 = a1 / np.linalg.norm(a1)
print("q1 =", q1)
print("Norm of q1 =", np.linalg.norm(q1))
```

```python
### The projection of a2 onto q1 is proj = <a2, q1>q1.

projection = np.dot(a2, q1) * q1
print("Projection of a2 onto q1 =", projection)
```

```python
### To create a vector orthogonal to q1, subtract the projection from a2. b2 = a2 - proj

b2 = a2 - projection
print("b2 =", b2)
```

```python
### The vector b2 is orthogonal to q1, but it is not yet a unit vector. To obtain q2, normalize b2. q2 = b2 / ||b2||

q2 = b2 / np.linalg.norm(b2)
print("q2 =", q2)
print("Norm of q2 =", np.linalg.norm(q2))
```

```python
### For q1 and q2 to form an orthogonal set: ||q1|| = 1, ||q2|| = 1, <q1, q2> = 0

print("Norm of q1 =", np.linalg.norm(q1))
print("Norm of q2 =", np.linalg.norm(q2))
print("Dot product of q1 and q2 =", np.dot(q1, q2))
```

```python
### The Gram Schmidt theorem states that q1 and q2 should span the same subspace as a1 and a2

A = np.column_stack((a1, a2))
Q = np.column_stack((q1, q2))

rank_A = np.linalg.matrix_rank(A)
rank_Q = np.linalg.matrix_rank(Q)

print("Rank of original vectors =", rank_A)
print("Rank of orthonomal vectors =", rank_Q)
```

1.2.4 Eigenvalues and eigenvectors

```python
### In this example, I will use the matrix A = [[2,1], [1,2]] and then verify that Ax = 𝜆x

A = np.array([[2, 1], [1, 2]], dtype=float)
print("Matrix A:")
print(A)
```

```python
### Each eigenvector should satisfy the equation Ax = 𝜆x

eigenvalues, eigenvectors = np.linalg.eig(A)
print("Eigenvalues:")
print(eigenvalues)
print("\nEigenvectors:")
print(eigenvectors)
```

```python
### The first eigenvalue is 𝜆1 = 3

lambda1 = eigenvalues[0]
x1 = eigenvectors[:, 0]
print("lambda1 =", lambda1)
print("x1 =", x1)
```

```python
Ax1 = A @ x1
lambda_x1 = lambda1 * x1
print("A x1 =", Ax1)
print("lambda1 x1 =", lambda_x1)
```

```python
### The values of Ax1 and 𝜆1x1 are equal. Therefore, x1 is an eigenvector of A corresponding to the eigenvalue 𝜆1 = 3. The second eigenvalue is 𝜆2 =1.

lambda2 = eigenvalues[1]
x2 = eigenvectors[:, 1]
print("lambda2 =", lambda2)
print("x2 =", x2)
```

```python
Ax2 = A @ x2
lambda_x2 = lambda2 * x2
print("A x2 =", Ax2)
print("lambda2 x2 =", lambda_x2)
```

```python
### Therefore Ax = 𝜆x was verified for both eigenvalue-eigenvector pairs.
```
