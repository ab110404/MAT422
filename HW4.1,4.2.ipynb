4.2. Graph and graph modeling
- Graph modeling is used to represent relationships between objects using nodes and edges.
- Graphs can be directed or undirected.
- In this example, I will create an undirected graph and calculate its shortest path and node degrees.

```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.Graph()
G.add_edges_from([
    ("A", "B"),
    ("A", "C"),
    ("B", "D"),
    ("C", "D"),
    ("D", "E")
])

nx.draw(
    G,
    with_labels=True,
    node_color="lightblue",
    node_size=1500,
    font_size=12
)

plt.title("Graph Modelining Example")
plt.show()
```

- A path is a sequence of edges connecting nodes.
- The shortest path is the path with the fewest edges between two nodes in an unweighted graph.
- The degree of a node is the number of edges connected to it.
- I will now calculate the shortest path from A to E and find the degree of each node.

```python
shortest_path = nx.shortest_path(G, "A", "E")
print("Shortest path:", shortest_path)
print("Path length:", len(shortest_path) - 1)
print("\nNode degrees:")
for node, degree in G.degree():
  print(node, ":", degree)
```

Conclusion:
- In this section, I created an undirected graph using Python.
- I used nodes and edges to represent connections and calculated the shortest path between two nodes.
- I also calculated node degrees to identify which nodes have the most connection.
- Node D had the highest degree with 3 connections.
This example demonstrates how graph modeling can be used to study relationships and networks.
